## 2. Variables & Data Types

A variable is just a name we give to a piece of data so we can refer back to it later, rather than retyping the value every time. In environmental science this is exactly how you would think about a single measurement: a temperature reading, a station name, a flag showing whether a sensor is active. Python figures out what kind of data you have given it automatically, so you never have to declare in advance that something is a number or a piece of text. This is called dynamic typing, and it is one of the reasons Python is a comfortable language to start with.


```python
# Store simple monitoring values, one variable per measurement
temperature = 23.5      # float (decimal number)
station = "Valletta"    # string (text)
is_active = True        # boolean (True/False)

# Print each variable alongside its type, so we can see how Python
# automatically worked out what kind of data each one holds
print("temperature:", temperature, "| type:", type(temperature))
print("station:", station, "| type:", type(station))
print("is_active:", is_active, "| type:", type(is_active))
```

**Explanation:**

`float`, `str`, and `bool` are three of Python's most common built-in types. A `float` is a number with a decimal point, a `str` is text (short for "string"), and a `bool` is a value that is either `True` or `False`. You can always check what type a variable holds by calling `type(x)` on it, which is a useful habit whenever you are not sure what kind of data you are working with, particularly once you start loading real datasets that might not be as clean as you expect.


## 3. Conditionals (Decisions)

Real environmental data is full of thresholds: a temperature above which we call a day "hot", a concentration above which a pollutant breaches a regulatory limit, a water level above which a river is at flood risk. Conditionals are how we teach Python to make exactly those kinds of decisions. The keywords `if`, `elif` ("else if"), and `else` let you branch your code so that different things happen depending on whether a condition is true or false.


```python
temperature = 28 # example measurement in °C

if temperature > 25:                 # test a condition
    print("Hot day")                 # runs only if the condition above is True
else:
    print("Mild")                    # runs only if the condition above is False
```

**Explanation:**

Python gives you the usual set of comparison operators to build these conditions: `>` (greater than), `>=` (greater than or equal to), `<` (less than), `<=` (less than or equal to), `==` (equal to, note the double equals sign, since a single `=` is used for assignment), and `!=` (not equal to). This is exactly the kind of logic you will use constantly for quality assurance work, for example flagging a reading if a nitrate concentration exceeds a regulatory limit.

## 4. Loops (Repetition)

Loops let you repeat a block of code automatically instead of copying and pasting it by hand. This matters enormously in data analysis, where you might need to apply the same check or calculation to hundreds or thousands of measurements. The `for` loop is the one you will reach for most often, and `range(n)` is a handy way to generate a simple sequence of whole numbers to loop over, running from 0 up to (but not including) `n`.


```python
for i in range(3):                  # range(3) yields 0, 1, 2 in turn
    print("Sample number:", i)      # this line runs once for each value of i, so 3 times total
```

**Explanation:**

`for` loops iterate over a sequence, meaning they take each item in turn and run the loop body once per item. This is exactly the pattern you would use to batch-process a folder of files or step through a list of records one by one.

Remember that Python starts counting its index from 0, not 1. This trips up almost everyone at first, so do not worry if it feels unnatural. It will become second nature with practice.


## 5. Lists (Ordered Collections)

A list is Python's way of holding an ordered collection of values, such as a series of monitoring stations or a week of rainfall totals. Lists are mutable, which means you can change them after creating them: add items, remove items, or update a value in place. This makes them a natural fit for datasets that grow or change as you work with them.


```python
stations = ["Marsaxlokk", "Gozo", "Valletta"]  # create a list of station names
print("First station:", stations[0])           # indexing starts at 0, so this is the first item
stations.append("Mellieha")                     # .append() adds a new item to the end of the list
print("All stations:", stations)                # the list now includes the newly appended station
```

**Explanation:**

Square brackets `[]` are how you define a list in Python, and how you access an item inside one by its position. The method `.append()` adds a new item to the end of an existing list. Remember that indexing starts at 0, so `stations[0]` gives you the first station in the list, not the second.

**Try changing the number inside `stations[ ]`, and see what output you get.**


## 6. Dictionaries (Key-Value Pairs)

A dictionary stores data as pairs of a key and a value, rather than as a simple ordered sequence like a list. This maps very naturally onto environmental readings, where you often have a set of named parameters, each with its own measured value, such as pH paired with 7.2 and dissolved oxygen paired with 6.5. Instead of remembering that "the third item in the list is dissolved oxygen", you look it up directly by name.


```python
readings = {"pH": 7.2, "DO": 6.5}        # create a dictionary of readings, keyed by parameter name
print("pH value:", readings["pH"])        # access a value by looking up its key
readings["Nitrate"] = 42                  # add a new key:value pair to the dictionary
print("All readings:", readings)          # the dictionary now includes the Nitrate entry too
```

**Explanation:**

Curly braces `{}` are how you define a dictionary in Python. You access a value using its key inside square brackets, for example `readings['pH']`, which is far more readable than trying to remember numeric positions the way you would with a list.


## 7. Indexing & Slicing (Lists)

Often you do not want a single value from a list, you want a subset of it, for example the first three days of a rainfall record or everything after the fourth reading. Slicing lets you extract exactly that kind of subset using the notation `[start:end]`, where the end position is excluded from the result. Negative indexes count backwards from the end of the list, so `-1` always refers to the last item.


```python
rainfall = [12, 8, 15, 20, 5, 0, 18]  # one rainfall value per day of the week

print("First value:", rainfall[0])     # index 0 is the first item, 12
print("Last value:", rainfall[-1])     # negative index -1 is the last item, 18

print("First three:", rainfall[0:3])   # slice from position 0 up to (not including) 3 -> [12, 8, 15]
print("From 4th on:", rainfall[3:])    # slice from position 3 to the end -> [20, 5, 0, 18]
print("Up to 5th:", rainfall[:5])      # slice from the start up to (not including) position 5 -> [12, 8, 15, 20, 5]
```

**Explanation:**

The start index in a slice is inclusive, meaning it is included in the result, while the end index is exclusive, meaning it is not. This is a common source of off-by-one confusion when you are starting out, so it is worth checking your slices against the actual output until the rule feels automatic. Slicing is extremely useful for pulling out a quick subset of a dataset before you run a full analysis on it.


## 8. Strings & Text Handling

Strings are how Python represents text, and in environmental datasets you will constantly be dealing with text: station names, sample codes, category labels. Being able to clean and manipulate strings reliably, for example standardising capitalisation or checking whether one piece of text is contained in another, is a basic but essential skill for tidying up messy real-world data.


```python
station = "Valletta"
print(station.upper())   # .upper() converts every letter to upper case -> 'VALLETTA'
print(station.lower())   # .lower() converts every letter to lower case -> 'valletta'
print(station[0:3])      # slicing a string works the same way as slicing a list -> 'Val'
print("Gozo" in station) # the "in" keyword checks whether one string sits inside another -> False
```

## 9. Functions (Reusable Mini-Programs)

A function is a named, reusable block of code that you define once and can then call as many times as you like. Functions exist to save you from repeating yourself. If you find yourself copying the same few lines of code over and over with only small changes, that is usually a sign you should be writing a function instead. This also makes your code far easier to read and to fix, since a correction only needs to be made in one place.


```python
def c_to_k(c):                 # define a function that takes one input, c (temperature in Celsius)
    """Convert Celsius to Kelvin."""   # docstring explains what the function does
    return c + 273.15                 # send the converted value back to whoever called the function

t_c = 25
t_k = c_to_k(t_c)                     # call the function with t_c as the argument
print("Temperature in Kelvin:", t_k)
```

**Explanation:**

The keyword `def` defines a function. The values you pass into it are called parameters, and inside the function body `return` sends a result back to whatever code called the function. The triple-quoted text directly under the function definition is called a docstring, and its job is to document what the function does for anyone reading or reusing your code later, including yourself in six months' time.


## 10. File Handling (Write & Read a Tiny Text File)

Sooner or later you will need to save results to disk and read data back in from a file, and this section shows the basic mechanics of doing that. We will write a small text file and then read it straight back. The `with` block is the standard, safe way to work with files in Python, because it automatically closes the file for you once you are done, even if something goes wrong inside the block.


```python
import os # built-in package for quick directory view
os.getcwd() # check the current working directory, i.e. where files will be read from and written to
```

```python
# Write a tiny text file
with open("data.txt", "w") as f:   # "w" mode creates the file (or overwrites it if it already exists)
    f.write("Sample1,23.5\n")      # \n starts a new line after this entry
    f.write("Sample2,19.8\n")

# Read it back
with open("data.txt", "r") as f:   # "r" mode opens the file for reading only
    contents = f.read()            # read the entire file into a single string

print(contents)
```

**Explanation:**

`open(path, mode)` opens a file at the given location. The mode `'w'` opens it for writing (creating the file if it does not exist, and overwriting it if it does), while `'r'` opens it for reading. Wrapping the call in a `with` block ensures the file is closed correctly once you leave the block, which avoids a whole class of bugs where files are left open and either lock other programs out or lose unsaved data. The `\n` character inside a string is not printed literally, it inserts a new line.


## 11. Light Use of NumPy (Arrays & Vectorised Math)

NumPy is the foundational numerical library in Python's scientific stack, and almost every other data tool you will use in this module, including pandas, is built on top of it. Its core object is the array, which behaves like a list of numbers but is far faster and lets you apply an operation to every element at once, instead of writing a loop yourself. This is called vectorisation, and it is the reason numerical work in Python is fast enough to handle large environmental datasets.


```python
import numpy as np

arr = np.array([12, 8, 15, 20, 5, 0, 18])  # create a NumPy array from a plain list
print("Third value:", arr[2])              # indexing works the same way as with a list
print("Slice 1..3:", arr[1:4])             # slicing works the same way as with a list too

print("Double:", arr * 2)                  # vectorised multiply: every element is doubled at once
print("Mask > 10:", arr > 10)              # boolean mask: True/False for each element, based on the comparison
```

**Explanation:**

Vectorised operations, such as multiplying every element of an array by 2 in a single line, are both more compact to write and much faster to run than writing an explicit loop over each element. Boolean masks, produced by a comparison like `arr > 10`, are a particularly powerful pattern: you get back an array of `True`/`False` values that you can then use to filter the original array, for example `arr[arr>10]` keeps only the values greater than 10.


## 12. Light Use of pandas (Tables)

pandas is the library you will use most often in this module for handling tabular data. Its central object, the DataFrame, behaves like a spreadsheet inside your code: rows and columns, with labels for both, and a huge set of built-in tools for selecting, filtering, and summarising data. Almost any environmental monitoring dataset, whether it is daily rainfall, weekly water quality samples, or a time series of sensor readings, fits naturally into this table structure.


```python
import pandas as pd

data = {"Day": ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"],
        "Rainfall": [12, 8, 15, 20, 5, 0, 18]}
df = pd.DataFrame(data)  # build a DataFrame (table) from the dictionary above

print("Full DataFrame:\n", df) # print the full dataframe
print("\nFirst 3 rows:\n", df.head(3))          # .head(3) previews just the first 3 rows
print("\nRainfall column:\n", df["Rainfall"])    # select a single column by its name
print("\nRows 0..1, column 1:\n", df.iloc[0:2, 1])  # select by numeric position: rows 0-1, column index 1
```

## 13. Descriptive Statistics (Quick Summary)

Before you do anything sophisticated with a dataset, you should always look at its basic descriptive statistics first: the mean, the median, and the standard deviation. These give you an immediate sense of where the data is centred and how spread out it is, and they are often enough to catch an obvious problem, such as a sensor that has drifted or a unit conversion that has gone wrong, before you waste time on deeper analysis.


```python
import statistics as stats

nitrates = [32, 45, 28, 50, 60, 42, 38]
print("Mean:", stats.mean(nitrates))       # the average value
print("Median:", stats.median(nitrates))   # the middle value when sorted
print("Std Dev:", stats.stdev(nitrates))   # standard deviation, a measure of spread
print("Min:", min(nitrates))               # the smallest value
print("Max:", max(nitrates))               # the largest value
```

**Explanation:**

Descriptive statistics are your first look at any dataset, and you should compute them as a matter of habit before moving on to anything more advanced. Use them to spot obvious issues, such as an implausible mean or a standard deviation far larger than you would expect, before you trust the data enough to build a model on top of it.


## 14. Generate a Tiny Synthetic Dataset (Air Quality)

To give everyone the same data to work with, without relying on a download that might fail or a file that might go missing, we will synthesise a week of hourly PM2.5 readings for a single monitoring station. PM2.5 refers to fine particulate matter with a diameter of 2.5 micrometres or less, a common air quality pollutant that follows a clear daily, or diurnal, pattern driven by traffic and atmospheric conditions. We will build that pattern ourselves and add some random noise on top, so the data behaves a little like a real sensor record.


```python
import numpy as np
import pandas as pd

np.random.seed(42)  # fixing the random seed means everyone gets exactly the same "random" numbers

hours = pd.date_range("2025-01-01", periods=24*7, freq="h")  # one full week of hourly timestamps
base = 25 + 10*np.sin(2*np.pi*(hours.hour)/24)               # a smooth diurnal (daily) pattern using a sine wave
noise = np.random.normal(0, 5, size=len(hours))              # random noise, one value per hour, to mimic real sensor variation

pm25 = np.clip(base + noise, 0, None)                        # add the noise to the base pattern, and clip so nothing goes below 0
```

```python
print("Base data:\n", base)    # the smooth diurnal pattern before noise was added
print("Added noise:\n", noise)  # the random noise we added on top of it
```

```python
print("Particulate Matter 2.5\u03bcm:\n\n", pm25) #\u03bcm is the unicode code for the micrometre symbol (µm)
```

```python
aq = pd.DataFrame({"datetime": hours, "PM25": pm25}) # combine the timestamps and readings into a dataframe
aq.set_index("datetime", inplace=True) # use the datetime column as the row index instead of 0, 1, 2, etc.
aq.head()
```

**Explanation:**

We simulate a simple diurnal cycle using a sine wave, then add random noise on top to make the data look more like a real, imperfect measurement rather than a perfectly smooth curve. `np.clip` prevents the result from going below zero, since a negative pollutant concentration would be physically meaningless. Generating the data this way means everyone in the class runs exactly the same workflow, without needing to download an external file.


## 15. Daily Means & Exceedance Flags

Hourly data is often more detail than you need for a summary, so a common step is to aggregate it down to a daily mean. Once you have that daily mean, you can compare it against a regulatory or guideline limit, here 35 micrograms per cubic metre for PM2.5, and flag any day that exceeds it. This exceedance-flagging pattern, aggregate then compare against a threshold, comes up constantly in environmental monitoring and quality assurance work.


```python
DAILY_LIMIT = 35  # the guideline limit we are checking each daily mean against, in µg/m³

daily = aq.resample("D").mean(numeric_only=True)  # group by calendar day and average the hourly readings
daily["exceeds"] = daily["PM25"] > DAILY_LIMIT     # True on any day the daily mean is above the limit
daily
```

**Explanation:**

`.resample('D')` groups the data by calendar day, using the datetime index we set earlier, in the same way you might group rows in a spreadsheet by date. We then compute the mean PM2.5 value for each day and create a new boolean column, `exceeds`, that is `True` on any day the daily mean is above the limit and `False` otherwise.

## 17. Mini Linear Regression (Rainfall vs. Nitrate)
A small predictive model to show the idea of regression. We generate tiny arrays and fit a line.

> Note: This is a *simple example* to illustrate the syntax and outputs.

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Example dataset
rainfall = [20,30,25,40,50,60,70]        # independent variable (mm)
nitrate  = [5, 7, 6, 9,11,12,15]         # dependent variable (mg/L)

X = np.array(rainfall).reshape(-1,1)     # scikit-learn expects a 2D array of inputs
y = np.array(nitrate)                    # the target values stay as a 1D array

model = LinearRegression()
model.fit(X, y)                             # fit the straight line to our data

print("Slope (m):", model.coef_[0])        # change in nitrate per 1 mm of extra rainfall
print("Intercept (c):", model.intercept_)  # predicted nitrate value when rainfall = 0
print("R²:", model.score(X, y))            # goodness of fit, from 0 (poor) to 1 (perfect)
```

**Explanation:**

Linear regression fits a straight line, `y = m·x + c`, through your data by finding the values of the slope `m` and intercept `c` that best describe the relationship between `x` and `y`. In scikit-learn, `coef_` gives you the fitted slope and `intercept_` gives you `c`, while `R²` tells you how well that line actually explains the variation in your data, on a scale from 0 (no relationship captured) to 1 (a perfect fit). We use a tiny, deliberately simple dataset here purely so the mechanics of fitting and reading the model are clear.


## 18. Worked QA/QC Example (Dictionary + Loop + Condition)

This example ties together everything from this practical into one small but realistic quality assurance task: given a set of nitrate readings at different sites, stored in a dictionary, loop through each site and flag it as exceeding a limit or not. This exact pattern, store data as key-value pairs, loop over it, apply a condition, is one you will reuse constantly once we move on to larger, real datasets later in the module.


```python
nitrate_map = {"A": 32, "B": 55, "C": 41, "D": 12}  # nitrate reading at each site
LIMIT = 50  # the threshold we are checking against

for site, val in nitrate_map.items():   # .items() gives us both the key (site) and the value (val) together
    if val > LIMIT:
        print(site, "EXCEEDS limit")
    else:
        print(site, "OK")
```
